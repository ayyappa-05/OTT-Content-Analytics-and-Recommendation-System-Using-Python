```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Platform', hue='type', data=df, palette='Set2')
plt.title('Content Type Breakdown by Platform')
plt.xlabel('Platform')
plt.ylabel('Count')
plt.show()
```

```python
# Filtering for recent decades to avoid squeezing the x-axis
recent_content = df[df['release_year'] >= 2000]

plt.figure(figsize=(14, 6))
sns.lineplot(data=recent_content.groupby(['release_year', 'type']).size().reset_index(name='count'), 
             x='release_year', y='count', hue='type', marker='o')
plt.title('Trend of Content Releases (2000 - Present)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
# Get top 10 most common ratings to keep it clean
top_ratings = df['rating'].value_counts().head(10).index
sns.countplot(x='rating', hue='Platform', data=df[df['rating'].isin(top_ratings)], palette='muted')
plt.title('Distribution of Content Ratings by Platform')
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Explode genres split by comma
genres = df['listed_in'].dropna().str.split(', ').explode()

# 2. Clean up duplicates (standardize plurals to singular/uniform terms)
genre_mapping = {
    'Dramas': 'Drama',
    'Comedies': 'Comedy',
    'Documentaries': 'Documentary',
    'International Movies': 'International Content',
    'International TV Shows': 'International Content'
}
genres = genres.replace(genre_mapping)

# 3. Get the top 15 clean counts
top_genres = genres.value_counts().head(15)

# 4. Plot without warnings
plt.figure(figsize=(12, 6))
sns.barplot(
    x=top_genres.values, 
    y=top_genres.index, 
    hue=top_genres.index,  # Maps y to hue to resolve the warning
    palette='viridis',
    legend=False           # Suppresses unnecessary legend box
)

plt.title('Top 15 Content Genres Across Platforms (Cleaned)')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```
